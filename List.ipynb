```python
#QUEUE >> follows first in first out
#it works in orderly sequential manner >> has two ends

from collections import deque
checkout = deque()
checkout.append("Cus1")
checkout.append("Cus2")
checkout.append("Cus3")
checkout.append("Cus4")
checkout.append("Cus5")
checkout
```

```python
while checkout: 
    customer = checkout.popleft()
    print("serving", customer)
```

```python
from queue import Queue
print_queue = Queue()
print_queue.put("Print the 1st pdf")            #put is similar to push
print_queue.put("Print the 2nd pdf")
print_queue.put("Print the 3rd pdf")
print_queue.put("Print the 4th pdf")
print_queue.put("Print the 5th pdf")

while not print_queue.empty():
    print_job = print_queue.get()           #get is similar to pop
    print("Printing", print_job)
```
