```python
#Dictionary comprehension
#Zip function

students = ["Ajay", "Sanjay", "Shyam"]
marks = [90, 80, 70]

for i in zip(students, marks):
    print(i)
```

```python
#store the above info in key value pair

students = ["Ajay", "Sanjay", "Shyam"]
marks = [90, 80, 70]
student_marks = {}
for student,marks in zip(students, marks):
    student_marks[student] = marks
student_marks
```

```python
#Using dictionary comprehension

students = ["Ajay", "Sanjay", "Shyam"]
marks = [90, 80, 70]
{student:marks for student, marks in zip(students, marks)}
```
